```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/sample_learning_log.csv")

df["date"] = pd.to_datetime(df["date"])

df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["date"],
    df["sleep_duration_hours"],
    marker="o",
    color="#4C78A8"
)

plt.title("Daily Sleep Duration")
plt.xlabel("Date")
plt.ylabel("Sleep Duration (Hours)")
plt.grid(alpha=0.3)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
